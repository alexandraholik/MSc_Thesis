```python
test_ids = set(pd.read_csv('data/test_set_100.csv')['article_id'].tolist())

full = pd.read_parquet('data/cc_news_matched.parquet')
full.index.name = 'article_id'
full = full.reset_index()

outside = full[~full['article_id'].isin(test_ids)].copy()
outside = outside[outside['best_sim'] < 0.97]

low_sim = outside[outside['best_sim'] < 0.55][
    ['article_id', 'title', 'headline_quote', 'best_body_sentence',
     'prev_sentence', 'next_sentence', 'best_sim', 'text']
].sample(30, random_state=42)

high_sim = outside[outside['best_sim'] > 0.80][
    ['article_id', 'title', 'headline_quote', 'best_body_sentence',
     'prev_sentence', 'next_sentence', 'best_sim', 'text']
].sample(30, random_state=42)

low_sim.to_csv('fewshot_candidates_contextomised.csv', index=False)
high_sim.to_csv('fewshot_candidates_modified.csv', index=False)
print("Done")
```
